## 2. Nuestra base de datos: contiene los datos que vamos a analizar

Para esta ayudantía trabajaremos con una base ficticia (creada mediante código) de una encuesta sobre **prácticas culturales, uso del tiempo y participación social**.




````r
# Base de datos de ejemplo
set.seed(123)

datos <- tibble(
  id = 1:100,
  edad = sample(18:70, 100, replace = TRUE),
  comuna = sample(c("Santiago", "Maipu", "Ñuñoa", "Recoleta", "Puente Alto"), 100, replace = TRUE),
  situacion_laboral = sample(c("Trabaja", "Estudia", "Trabaja y estudia", "No trabaja"), 100, replace = TRUE),
  horas_trabajo = sample(0:50, 100, replace = TRUE),
  horas_estudio = sample(0:35, 100, replace = TRUE),
  participacion_organizacion = sample(0:5, 100, replace = TRUE),
  satisfaccion_vida = sample(1:10, 100, replace = TRUE)
)
````

## Exploración de datos



````r
glimpse(datos)
````

````r
dim(datos)
````

# 3. Seleccionar variables → `select()`

**Pregunta:** ¿Cómo completamos este código si queremos explorar la posibles vinculaciones entre edad, situación laboral y nivel de participación en alguna(s) organización(es)?

````r
# Completa el código:
datos %>%
  select(edad, situacion_laboral, participacion_organizacion)
````

# 4. Filtrarar casos → `filter()`

**Pregunta:** ¿Cómo podemos quedarnos solamente con las personas menores de 30 años?

````r
datos %>%
  filter(edad < 30)
````

### ¿Qué ocurrirá con el número de filas?

- Aumentará
- Disminuirá
- Se mantendrá igual

## 5. Combinar `filter()` + `select()`

Ahora queremos estudiar **solamente a las personas menores de 30 años**, pero mirando únicamente su edad, situación laboral y nivel de participación en alguna organización(es).

````r
datos %>%
  filter(edad < 30) %>%
  select(edad, situacion_laboral, participacion_organizacion)
````

### Operadores lógicos

Podemos combinar condiciones con `&` (Y) y `|` (O).

Por ejemplo: personas mayores de 25 años **Y** que viven en Santiago.

````r
datos %>%
  filter(edad > 25 & comuna == "Santiago")
````

````r
datos %>%
  filter(edad > 18 & edad < 30 & situacion_laboral == "Trabaja y estudia")
````

# 6. Crear/modificar variables → `mutate()`

Ahora pasamos de **seleccionar y filtrar** a **transformar los datos**.

Si tenemos horas de trabajo y horas de estudio, podemos crear una variable que aproxime las horas de tiempo libre (si lo contemplamos como tiempo no trabajado ni estudiado) trabajo/estudio disponibles en un día

````r
datos %>%
  mutate(
    horas_disponibles = 24 - horas_trabajo - horas_estudio
  )
````

### 📌 Pregunta

Queremos crear una variable que indique si una persona tiene **participación alta** en alguna organización(es), considerando alta una puntuación igual o superior a 3.

¿Qué código escribirías?

````r
# Intenta antes de mirar la solución
datos %>%
  mutate(
    participacion_alta = participacion_organizacion >= 3
  )
````

# 7. Ordenar casos → `arrange()`

Queremos encontrar primero a quienes tienen **más horas de trabajo**.

````r
datos %>%
  arrange(desc(horas_trabajo))
````

### 🔎 ¿Qué diferencia hay?

Compara mentalmente:

```r
arrange(horas_trabajo)
```

y

```r
arrange(desc(horas_trabajo))
```


# 8. Contar casos → `count()`

**Pregunta:** ¿Cómo se distribuyen las personas de nuestra muestra según situación laboral?

````r
datos %>%
  count(situacion_laboral)
````

### Otra pregunta

¿Cómo se distribuye la participación en organizaciones?

````r
datos %>%
  count(participacion_organizacion, sort = TRUE)
````
